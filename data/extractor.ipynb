# Data Extraction from Espncricinfo

```python
import requests
import pandas as pd
from io import StringIO
```

```python
url = 'https://stats.espncricinfo.com/ci/engine/player/253802.html?class=2;home_or_away=1;home_or_away=2;home_or_away=3;result=1;result=2;result=3;result=5;template=results;type=batting;view=innings;wrappertype=print'
```

```python
headers={
    'User-Agent': 'Mozilla/5.0'
}
```

```python
res = requests.get(url,headers=headers)
```

```python
tables = pd.read_html(StringIO(res.text))
```

```python
tables
```

```python
inn_df = tables[3]
```

```python
inn_df = inn_df.drop(columns=['Unnamed: 9'])
```

```python
inn_df.to_csv("virat_odi_data.csv", index=False)
```
